```python
from pycocotools.coco import COCO

ann_file = "./ground_truth/train/instances_default_train.json"
coco = COCO(ann_file)

cats = coco.loadCats(coco.getCatIds())
ann_classes = [c["name"] for c in cats]
ann_ids = [c["id"] for c in cats]

print("Annotation class order (category_id → name):")
for cid, name in zip(ann_ids, ann_classes):
    print(cid, "→", name)

print("\nAnnotation class list (in file order):", ann_classes)
print("Annotation category_ids:", ann_ids)
```
